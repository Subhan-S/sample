```python
import pandas as pd
```

 Load the dataset into a DataFrame

```python
data = pd.read_csv('https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_month.csv')
df = pd.DataFrame(data)
df
```

 Identify number of records and unique locations

```python
# number of records
num_records = len(df)
print(f"Number of records: {num_records}")

# number of unique locations
num_unique_locations = df['place'].nunique()
print(f"Number of unique locations: {num_unique_locations}")
```

Print top 5 rows and column names

```python
print("Top 5 rows:")
print(df.head().to_markdown(index=False, numalign="left", stralign="left"))

print("\nColumn names:")
print(df.columns.tolist())
```

 Convert time to datetime

```python
import pandas as pd
df['time'] = pd.to_datetime(df['time'])
print("\nDataFrame with 'time' column converted to datetime:")
print(df.head().to_markdown(index=False, numalign="left", stralign="left"))
```

Drop records with missing latitude, longitude, mag

```python
df_cleaned = df.dropna(subset=['latitude', 'longitude', 'mag'])

print("\nDataFrame after dropping records with missing latitude, longitude, or mag:")
print(df_cleaned.head().to_markdown(index=False, numalign="left", stralign="left"))
print(f"\nNumber of records after dropping missing values: {len(df_cleaned)}")
```

Filter only earthquakes with magnitude >= 4.0

```python
df_strong_earthquakes = df_cleaned[df_cleaned['mag'] >= 4.0]

print("\nDataFrame filtered for earthquakes with magnitude >= 4.0:")
print(df_strong_earthquakes.head().to_markdown(index=False, numalign="left", stralign="left"))
print(f"\nNumber of strong earthquakes (mag >= 4.0): {len(df_strong_earthquakes)}")
```

 Add a column day_of_week from time

```python
df_strong_earthquakes['day_of_week'] = df_strong_earthquakes['time'].dt.day_name()
print("\nDataFrame with 'day_of_week' column:")
print(df_strong_earthquakes.head().to_markdown(index=False, numalign="left", stralign="left"))
```

Create a column severity_level based on magnitude:
○   <4.0: "Low"

○   4.0 - 6.0: "Moderate"

○   6.0+: "High"

```python
def get_severity_level(magnitude):
    if magnitude < 4.0:
        return "Low"
    elif 4.0 <= magnitude < 6.0:
        return "Moderate"
    else:
        return "High"

df_strong_earthquakes['severity_level'] = df_strong_earthquakes['mag'].apply(get_severity_level)

print("\nDataFrame with 'severity_level' column:")
print(df_strong_earthquakes.head().to_markdown(index=False, numalign="left", stralign="left"))
```

Count number of earthquakes per place

```python
earthquake_counts_by_place = df_strong_earthquakes['place'].value_counts()

print("\nNumber of earthquakes per place (for strong earthquakes):")
print(earthquake_counts_by_place.head().to_markdown(numalign="left", stralign="left"))
```

Compute average magnitude and max depth per day

```python
daily_summary = df_strong_earthquakes.groupby(df_strong_earthquakes['time'].dt.date).agg(
    average_magnitude=('mag', 'mean'),
    max_depth=('depth', 'max')
)

print("\nAverage magnitude and max depth per day:")
print(daily_summary.head().to_markdown(numalign="left", stralign="left"))
```

Save:

○   Cleaned dataset to cleaned_earthquakes.csv

○   Summary data to earthquake_summary.csv

```python
df_cleaned.to_csv('cleaned_earthquakes.csv', index=False)
daily_summary.to_csv('earthquake_summary.csv')

print("Cleaned dataset saved to cleaned_earthquakes.csv")
print("Summary data saved to earthquake_summary.csv")
```

Store into SQLite table

```python
import sqlite3
conn = sqlite3.connect('earthquakes.db')
df_cleaned.to_sql('earthquakes', conn, if_exists='replace', index=False)
conn.close()
print("Cleaned dataset stored in 'earthquakes' table in 'earthquakes.db'")
```
